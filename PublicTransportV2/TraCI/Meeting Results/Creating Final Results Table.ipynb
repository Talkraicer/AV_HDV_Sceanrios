```python
import pandas as pd
from pandas import MultiIndex
```

# Preparing LeftCompDaily

```python
av_rates = ["av0.1", "av0.2", "av0.3", "av0.4", "av0.6", "av0.8"]
```

```python
final_LeftCompDaily = pd.read_pickle("../results_csvs/scenarios_LeftCompDaily.pkl")
final_LeftCompDaily = final_LeftCompDaily["TOTAL"].drop(columns=["av0.5"])
final_LeftCompDaily.columns = MultiIndex.from_product([["LeftCompDaily"], final_LeftCompDaily.columns])
final_LeftCompDaily.dropna(inplace=True)
```

```python
final_LeftCompDaily
```

# Preparing LeftCompScenarios

```python
final_LeftCompScenarios = pd.read_pickle("../results_csvs/scenarios_LeftCompScenarios.pkl")
final_LeftCompScenarios = final_LeftCompScenarios.drop(columns=["TOTAL"], level=0)
final_LeftCompScenarios.dropna(inplace=True)
```

```python
final_LeftCompScenarios
```

# Preparing ClosedLeftCompScenarios

```python
ClosedLeftCompScenarios = pd.read_pickle("../results_csvs/scenarios_ClosedLeftCompScenarios.pkl")
ClosedLeftCompScenarios = ClosedLeftCompScenarios.drop(columns=["TOTAL"], level=0)
ClosedLeftCompScenarios.dropna(inplace=True)
```

```python
new_CloseLeftCompScenariosColumns = [("Closed_"+col[0], col[1]) for col in ClosedLeftCompScenarios.columns]
final_ClosedLeftCompScenarios = ClosedLeftCompScenarios.copy()
final_ClosedLeftCompScenarios.columns = MultiIndex.from_tuples(new_CloseLeftCompScenariosColumns)
```

```python
final_ClosedLeftCompScenarios
```

# Creating the final results table

```python
final_results_index =  ["NO PTL", "BUS-ONLY PTL", "Best Daily HOV policy", "Best HOV policy per scenario","Best Daily StaticNumFL", "Best StaticNumFL per scenario", "Best Daily Dynamic Policy", "Best Dynamic Policy per scenario"]
LeftCompDaily_empty = pd.DataFrame(columns=final_LeftCompDaily.columns, index=final_results_index)
LeftCompScenarios_empty = pd.DataFrame(columns=final_LeftCompScenarios.columns, index=final_results_index)
ClosedLeftCompScenarios_empty = pd.DataFrame(columns=final_ClosedLeftCompScenarios.columns, index=final_results_index)
final_results = pd.concat([LeftCompDaily_empty, LeftCompScenarios_empty, ClosedLeftCompScenarios_empty], axis=1)
```

```python
final_results
```

# adding no ptl and bus only ptl

```python
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["NO PTL",col] = final_LeftCompScenarios.loc["Plus_1",col]
        final_results.loc["BUS-ONLY PTL",col] = final_LeftCompScenarios.loc["Nothing",col]
    elif col in final_LeftCompDaily.columns:
        final_results.loc["NO PTL",col] = final_LeftCompDaily.loc["Plus_1",col]
        final_results.loc["BUS-ONLY PTL",col] = final_LeftCompDaily.loc["Nothing",col]
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["NO PTL",col] = final_ClosedLeftCompScenarios.loc["Plus_1",col]
        final_results.loc["BUS-ONLY PTL",col] = final_ClosedLeftCompScenarios.loc["Nothing",col]
```

```python
final_results
```

# Adding best daily HOV policy

```python
def row_with_max_min_count(df):
    min_values = df.min()
    is_min = df.eq(min_values)
    min_count_per_row = is_min.sum(axis=1)
    row_with_most_min_values = min_count_per_row.idxmax()
    return row_with_most_min_values
```

```python
HOVs = ["Plus_1", "Plus_2", "Plus_3", "Plus_4", "Plus_5"]
LeftCompDaily_HOVs =  final_LeftCompDaily.loc[HOVs]
LeftCompDaily_HOVs_idx = row_with_max_min_count(LeftCompDaily_HOVs)
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["Best Daily HOV policy",col] = final_LeftCompScenarios.loc[LeftCompDaily_HOVs_idx,col]
    elif col in final_LeftCompDaily.columns:
        final_results.loc["Best Daily HOV policy",col] = final_LeftCompDaily.loc[LeftCompDaily_HOVs_idx,col]
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["Best Daily HOV policy",col] = final_ClosedLeftCompScenarios.loc[LeftCompDaily_HOVs_idx,col]
```

```python
final_results
```

# Adding best HOV policy per scenario

```python
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["Best HOV policy per scenario",col] = final_LeftCompScenarios[col].loc[HOVs].min()
    elif col in final_LeftCompDaily.columns:
        final_results.loc["Best HOV policy per scenario",col] = final_LeftCompDaily[col].loc[HOVs].min()
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["Best HOV policy per scenario",col] = final_ClosedLeftCompScenarios[col].loc[HOVs].min()
```

```python
final_results
```

# Adding best daily StaticNumFL

```python
StaticNumFLs = [f"StaticNumPassFL_{i}" for i in range(1,6)]
LeftCompDaily_StaticNumFLs =  final_LeftCompDaily.loc[StaticNumFLs]
LeftCompDaily_StaticNumFLs_idx = row_with_max_min_count(LeftCompDaily_StaticNumFLs)
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["Best Daily StaticNumFL",col] = final_LeftCompScenarios.loc[LeftCompDaily_StaticNumFLs_idx,col]
    elif col in final_LeftCompDaily.columns:
        final_results.loc["Best Daily StaticNumFL",col] = final_LeftCompDaily.loc[LeftCompDaily_StaticNumFLs_idx,col]
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["Best Daily StaticNumFL",col] = final_ClosedLeftCompScenarios.loc[LeftCompDaily_StaticNumFLs_idx,col]
```

```python
final_results
```

# Adding best StaticNumFL per scenario

```python
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["Best StaticNumFL per scenario",col] = final_LeftCompScenarios[col].loc[StaticNumFLs].min()
    elif col in final_LeftCompDaily.columns:
        final_results.loc["Best StaticNumFL per scenario",col] = final_LeftCompDaily[col].loc[StaticNumFLs].min()
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["Best StaticNumFL per scenario",col] = final_ClosedLeftCompScenarios[col].loc[StaticNumFLs].min()
```

```python
final_results
```

# Adding best daily Dynamic Policy

```python
DynamicPolicies = list(set(final_LeftCompScenarios.index.tolist())-set(HOVs+StaticNumFLs+["Nothing"]))
LeftCompDaily_DynamicPolicies =  final_LeftCompDaily.loc[DynamicPolicies]
LeftCompDaily_DynamicPolicies_idx = row_with_max_min_count(LeftCompDaily_DynamicPolicies)
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["Best Daily Dynamic Policy",col] = final_LeftCompScenarios.loc[LeftCompDaily_DynamicPolicies_idx,col]
    elif col in final_LeftCompDaily.columns:
        final_results.loc["Best Daily Dynamic Policy",col] = final_LeftCompDaily.loc[LeftCompDaily_DynamicPolicies_idx,col]
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["Best Daily Dynamic Policy",col] = final_ClosedLeftCompScenarios.loc[LeftCompDaily_DynamicPolicies_idx,col]
```

# Adding best Dynamic Policy per scenario

```python
for col in final_results.columns:
    if col in final_LeftCompScenarios.columns:
        final_results.loc["Best Dynamic Policy per scenario",col] = final_LeftCompScenarios[col].loc[DynamicPolicies].min()
    elif col in final_LeftCompDaily.columns:
        final_results.loc["Best Dynamic Policy per scenario",col] = final_LeftCompDaily[col].loc[DynamicPolicies].min()
    elif col in final_ClosedLeftCompScenarios.columns:
        final_results.loc["Best Dynamic Policy per scenario",col] = final_ClosedLeftCompScenarios[col].loc[DynamicPolicies].min()
```

```python
# Highlight the min in each row
def highlight_min(s):
    # filter out indices that contain HOV
    s_without = s[~s.index.str.contains("HOV")]
    is_min = s == s.min()
    is_min_without = s == s_without.min()
    styles = []
    # set to bold and yellow if it is the min
    for v,u in zip(is_min, is_min_without):
        style = ''
        if v and u:
            style = 'background-color: yellow; font-weight: bold'
        elif v:
            style = 'background-color: yellow'
        elif u:
            style = 'font-weight: bold' 
        styles.append(style)
    return styles
final_results.style.apply(highlight_min, axis=0).to_excel("final_results.xlsx")
```
